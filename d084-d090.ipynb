```python
dado = []
nomepeso = []
pesado = leve = 0
while True:
    dado.append(str(input("Nome= ")))
    dado.append(float(input("Peso= ")))
    if len(nomepeso) == 0:
        pesado = leve = dado[1]
    else:
        if dado[1] > pesado:
            pesado = dado[1]
        if dado[1] < leve:
            leve = dado[1]
    nomepeso.append(dado[:])
    dado.clear()

    r = str(input("Quer continuar? = "))
    if r in "Nn":
        break

print(f"Ao todo, foram cadastrados {len(nomepeso)} pessoas")#esse len ja da a quantidade de termos em nomepeso 
print(f"O mais pesado tem {pesado} quilos e seu nome é ", end="")
for p in nomepeso:
    if p[1] == pesado:
        print(f"{p[0]}")
print(f"O mais leve tem {leve} quilos e seu nome é ", end="")
for p in nomepeso:
    if p[1] == leve:
        print(f"{p[0]}", end="")

```
